```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 2, 5],
    'age':         [25, np.nan, 35, np.nan, np.nan, 40],
    'city':        ['Mumbai', 'Delhi', np.nan, 'Mumbai', 'Delhi', np.nan],
    'gender':      ['Male', 'Female', 'Female', 'Male', 'Female', 'Male'],
    'annual_income': [40000, 80000, 60000, 120000, 80000, 95000]
})
df
```

# Task1 - Clean the data

```python
df['age'] = df['age'].fillna(df['age'].median())
df['city'] = df['city'].fillna(df['city'].mode()[0])
```

```python
df.duplicated().sum()
```

```python
df[df.duplicated()]
```

```python
df.drop_duplicates(inplace=True)
df
```

# Task 2 - Encode Categorical columns

```python
df = pd.get_dummies(df, columns=['city'], drop_first=True) # one hot encoding for city column
df
```

```python
df['gender'] = df['gender'].map({'Male': 1, 'Female': 0})
print("The resulting DataFrame is:")
print(df)
```

# Task 3- Scale Numerical Features

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

df_minmax = df.copy()
df_std = df.copy()
cols = ['age','annual_income']
minmax = MinMaxScaler()
df_minmax[cols] = minmax.fit_transform(df_minmax[cols])
print("DataFrame after Min-Max Scaling:")
print(df_minmax)

std = StandardScaler()
df_std[cols] = std.fit_transform(df_std[cols])
print("DataFrame after Standardization:")
print(df_std)
```

Min-Max Scaling rescales data into a fixed range (usually 0 to 1). It is preferred when you need bounded values (e.g., neural networks, distance-based models like KNN).
Standardization transforms data to have mean = 0 and standard deviation = 1. It is preferred when data follows a normal distribution or for models like Linear Regression and SVM.
